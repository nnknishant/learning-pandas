```python
import pandas as pd
```

```python

df_employee = {
    "Emp_ID": [101, 102, 103, 104, 105, 106],
    "Name": ["Rahul", "Priya", "Amit", "Neha", "Ravi", "Anjali"],
    "Department": ["IT", "HR", "IT", "Finance", "Sales", "Sales"],
    "Salary": [65000, 55000, 72000, 60000, 48000, 75000],
    "Experience": [3, 2, 5, 4, 1, 6]
}

```

```python
df = pd.DataFrame(df_employee)
```

```python
df
```

```python
# Question 1: Find employees whose salary is greater than ₹60,000

salary = df[df['Salary'] > 60000]
salary
```

```python
# Question 2: Create a new column Annual_Salary.

df['annual_salary'] = df['Salary'] * 12
```

```python
df
```

```python
df
```

```python
# Question 3: Find the employee with the highest salary.

highest_salarys = df.loc[df['Salary'].idxmax()]
```
