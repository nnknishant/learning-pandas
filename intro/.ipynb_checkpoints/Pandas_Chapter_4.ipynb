2 Filtering & Conditional Selection

```python

# Get all employees in the IT department earning more than 65,000.

df_filtered = df[(df['Department'] == 'IT') & (df['Salary'] > 65000)]
df_filtered
```

```python
# Get all employees in the Marketing department earning more than 40,000.
9
df_filtered2 =df[(df['Department'] == 'Marketing') & (df['Salary'] > 40000)]
df_filtered2
```

3. Sorting & Ranking


```python
# Find the top 3 highest-paid employees

top3_salary = df.nlargest(3,columns='Salary')
top3_salary
```

```python
# How do you rank employees by salary within each department?
df_ranked = df.groupby('Department')['Salary'].rank(ascending=False)
df_ranked
```

4. GroupBy & Aggregation


```python
# Calculate average salary by department.

df_avg_salary = df.groupby('Department')['Salary'].mean().round(0)
df_avg_salary
```
