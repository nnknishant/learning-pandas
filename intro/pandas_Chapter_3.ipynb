```python
    import pandas as pd
    import numpy as np
```

```python
# Sample Interview Dataset: Employee Performance Data
data = {
    "Emp_ID":[101,102,103,104,105,106,107,108,109,110],
    "Name": ["Anna", "Ben", "Charlie", "Diana", "Evan", "Fiona", "George", "Hannah", "Ian", "Jack"],
    "Department": ["Sales", "HR", "Sales", "IT", "Sales", "HR", "IT", "Sales", "IT", "Sales"],
    "Salary": [55000, 60000, 72000, 85000, 50000, np.nan, 90000, 68000, 82000, 72000],
    "Sales_Target_Achieved": [True, False, True, True, False, False, True, True, False, True],
    "Region": ["North", "South", "North", "West", "East", "South", "West", "North", "East", "North"]
}
```

```python
df = pd.DataFrame(data)
```

```python
df
```

```python
# Multiple condition filtering (Note the parentheses around each condition)

# Filter by department and sales more than 50k

df_sales = df[(df['Department'] == "Sales") & (df["Salary"] > 50000)]
```

```python
df_sales
```

```python
# Name starting by H

df_names = df[df['Name'].str.startswith('H')]
```

```python
df_names    
```

```python
# Group BY multiple condion

df_groupby = df.groupby(['Department', 'Region']).agg({'Salary':['mean','max']})
```

```python
df_groupby
```

```python
# Group by by single condition
df_groupby2 = df.groupby(['Department']).sum().head(2)
```

```python
df_groupby2 
```

```python
# Group by with some multiple and single conditions.

df_groupby3 = df.groupby(['Department', 'Region']).agg({'Salary':['mean', 'max', 'min','sum']})
```

```python
df_groupby3
```

```python
# Filter by IT or HR department sales higher than 85000

df_filtered = df[(df['Department'] == 'IT') | (df['Department'] == 'HR') & (df['Salary'] >85000)]
```
