```python
import pandas as pd
import numpy as np
```

```python
data = {
    'EmployeeID': [101, 102, 103, 104, 105, 106, 107, 108, 109, 110],
    'Name': ['Alice Smith', 'Bob Jones', 'Carla Diaz', 'David Lee', 'Eva Kim',
             'Frank Wu', 'Grace Patel', 'Henry Ford', 'Ivy Chen', 'Jack Ryan'],
    'Department': ['Sales', 'Sales', 'Marketing', 'IT', 'Marketing',
                   'IT', 'Sales', 'HR', 'Marketing', 'IT'],
    'Region': ['East', 'West', 'East', 'West', 'East',
               'South', 'South', 'East', 'West', 'South'],
    'Sales': [45000, 54000, np.nan, 30000, 62000,
              28000, 71000, np.nan, 39000, 48000],
    'Rating': [4.5, 3.8, 4.1, np.nan, 4.9, 3.2, 4.7, 3.9, 4.0, 4.4]}
```

```python
# Find sales which grater than 54000

high_sales = df[df['Sales'] > 54000]
high_sales
```

```python
# Filer with Ratings which higher than 4.1 and sales > 45000

high_rating  =  df[(df['Rating']> 4.1) & (df['Sales'] > 45000)]
print(high_rating)
```

```python
# Find sales department with amount greater than 54000

sales_dept = df[(df['Department'] == 'Sales') & (df['Sales'] >54000)]
print(sales_dept)
print(sales_dept[['Name', 'Department', 'Sales']])
```

```python
# How  would you clean nulls here?

sales_null = df[df['Sales'].isnull()]
print(sales_null)
```

```python
# Filter  by Region and sales

region_filter = df[(df['Region'] == 'East') & (df['Sales'] > 54000)]

region_filter
```
